# Phishing URL Checker

**Objective:** Use simple rules to flag common warning signs in a URL. This is an educational demo, not a definitive phishing detector.

### How it works
1. Read a URL from the user.
2. Parse the URL and inspect basic features.
3. Add warning points for selected indicators.
4. Display a risk band and the reasons.


```python
from urllib.parse import urlparse

SUSPICIOUS_WORDS = (
    'login', 'verify', 'update', 'secure', 'account', 'banking',
    'password', 'confirm', 'signin', 'wallet', 'gift'
)

def check_url(url):
    url = url.strip()
    reasons = []
    score = 0
    if not url:
        return {'url': url, 'score': 0, 'risk': 'Invalid input',
                'reasons': ['Please enter a URL.']}
    candidate = url if '://' in url else 'http://' + url
    parsed = urlparse(candidate)
    host = (parsed.hostname or '').lower()
    if parsed.scheme != 'https':
        score += 1; reasons.append('Does not use HTTPS')
    if not host or '.' not in host:
        score += 2; reasons.append('Hostname looks incomplete')
    if '@' in url:
        score += 2; reasons.append('Contains @ symbol, which can disguise the destination')
    if len(url) > 75:
        score += 1; reasons.append('Unusually long URL')
    if host.count('-') >= 2:
        score += 1; reasons.append('Hostname contains several hyphens')
    if any(word in url.lower() for word in SUSPICIOUS_WORDS):
        score += 1; reasons.append('Contains words often seen in urgent account messages')
    if sum(ch.isdigit() for ch in host) >= 5:
        score += 1; reasons.append('Hostname contains many digits')
    if score >= 4:
        risk = 'High warning signs'
    elif score >= 2:
        risk = 'Some warning signs'
    else:
        risk = 'Few warning signs detected'
    if not reasons:
        reasons.append('No basic warning signs detected by these rules')
    return {'url': url, 'score': score, 'risk': risk, 'reasons': reasons}

url = input('Enter a URL to check: ')
result = check_url(url)
print('\nAssessment:', result['risk'])
print('Warning score:', result['score'])
print('Reasons:')
for reason in result['reasons']:
    print('-', reason)
print('\nReminder: this basic tool cannot guarantee that a website is safe.')
```

### Test cases
Try entering `http://account-verify-login.example.com` or `https://www.wikipedia.org`.

### Limitations
- Rules may produce false positives and false negatives.
- HTTPS alone does not prove a site is trustworthy.
- The program does not visit URLs, download content, or send data to a server.
